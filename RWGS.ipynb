```python
import matplotlib.pyplot as plt
import numpy as np

T = np.linspace(300, 1400, 100)

def equilibrium_constant(T):
    K_eq = np.exp(-13.148 + 1.077 * np.log(T) + 5.44e-4 * T - 1.125e-7 * T**2 + 5693.5 / T - 49170 / T**2)
    return K_eq

# Calculate equilibrium constant values
K_eq = equilibrium_constant(T)

# Create the plot
plt.figure(figsize=(6, 4))
plt.plot(T, K_eq, linewidth=0.8, color='blue')
plt.xlabel('Temperature (K)', fontsize=8)
plt.ylabel('Equilibrium Constant', fontsize=8)
plt.title('Equilibrium Constant vs Temperature', fontsize=10)
plt.grid(True, alpha=0.5)
plt.xlim(500, 1000)
plt.ylim(0, 150)  # Set y-axis limit to match your image
plt.tight_layout()
plt.show()
```

```python
# Calculate extent of reaction from K(T)
# For equimolar feed: ξ = √K / (1 + √K)
xi = np.sqrt(1 / K_eq) / (1 + np.sqrt(1 / K_eq))

# Plot reaction extent vs temperature
plt.figure(figsize=(6, 4))
plt.plot(T, xi, linewidth=0.8, color='blue')
plt.xlabel('Temperature (K)', fontsize=8)
plt.ylabel('Reaction Extent (ξ)', fontsize=8)
plt.title('Reaction Extent vs Temperature', fontsize=10)
plt.grid(True, alpha=0.5)
plt.xlim(500, 1000)
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
